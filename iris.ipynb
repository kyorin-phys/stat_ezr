```r
# 初回のみパッケージをインストールしてください
# install.packages("ggplot2")
# install.packages("GGally")

library(ggplot2)
library(GGally)

# irisデータを使って散布図行列を作成
ggpairs(
  data = iris,
  columns = 1:4,                              # プロットする列（1〜4列目の数値データ）
  mapping = aes(color = Species, alpha = 0.6), # 種別で色分け、透明度を設定
  title = "Irisデータの散布図行列 (Species別)"
) +
  theme_minimal() +                           # ggplot2のテーマを適用可能
  theme(plot.title = element_text(face = "bold", size = 14))
```
